```python
# CORD-19 Analysis Notebook

# --------------------------
# Part 1: Data Loading
# --------------------------
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset (adjust path as needed)
df = pd.read_csv("../data/metadata.csv", low_memory=False)

# Basic exploration
print("Shape:", df.shape)
print(df.info())
print(df.head())

# --------------------------
# Part 2: Data Cleaning
# --------------------------
# Convert publish_time to datetime
df['publish_time'] = pd.to_datetime(df['publish_time'], errors='coerce')

# Extract year
df['year'] = df['publish_time'].dt.year

# Abstract word count
df['abstract_word_count'] = df['abstract'].fillna("").apply(lambda x: len(x.split()))

# Handle missing titles/journals
df['title'] = df['title'].fillna("Unknown Title")
df['journal'] = df['journal'].fillna("Unknown Journal")

# --------------------------
# Part 3: Analysis & Visualization
# --------------------------
# Publications per year
year_counts = df['year'].value_counts().sort_index()
plt.figure(figsize=(8,4))
plt.bar(year_counts.index, year_counts.values)
plt.title("Publications by Year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()

# Top journals
top_journals = df['journal'].value_counts().head(10)
plt.figure(figsize=(8,5))
sns.barplot(y=top_journals.index, x=top_journals.values)
plt.title("Top 10 Journals")
plt.xlabel("Number of Papers")
plt.ylabel("Journal")
plt.show()

# Word frequency in titles
from collections import Counter
import re

titles = " ".join(df['title'].dropna().tolist()).lower()
words = re.findall(r'\b[a-z]{4,}\b', titles)
common_words = Counter(words).most_common(20)

common_words_df = pd.DataFrame(common_words, columns=["word","count"])
plt.figure(figsize=(10,5))
sns.barplot(x="count", y="word", data=common_words_df)
plt.title("Most Common Words in Titles")
plt.show()
```
